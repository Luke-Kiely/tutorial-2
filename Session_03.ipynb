### TUPLE (Immutable Metadata): Fixed identifiers that must never change

A tuple is an immutable data structure in Python, meaning that once it is created, its contents cannot be altered. This makes tuples ideal for storing fixed identifiers and metadata that should remain constant throughout the lifecycle of a program.
A tuple is defined using parentheses `()` and can hold multiple items separated by commas of different data types, such as strings, integers, and floats. A tuple is just a collection of items that are grouped together, and it can be used to represent a single entity or a set of related values. Its elements can be accessed using indexing.

Scenario:Based on the invoice above, we can define a tuple to hold immutable metadata about the transaction. This tuple will include the invoice number, date of issue, exporter and importer details, and shipping information.

```python
invoice_metadata = (
    "INV-2026-8849A",  # Invoice Number
    "02-SEP-2026",     # Date of Issue
    "COOXUPÉ COOP",    # Exporter Name
    "DUBLIN ROASTERS & IMPORTERS LTD",  # Importer Name
    "FOB Santos Port",  # Incoterm
    "M/V SANTA MARIA",  # Vessel Name
    "MSCUBR827104921",  # Bill of Lading Reference
    "USD"              # Currency
)
# and a print statement to display the tuple
print("Invoice Metadata:", invoice_metadata)
```

We can also test the immutability of the tuple by trying to change one of its elements. This will raise an error, demonstrating that tuples cannot be modified after creation.

```python
# Attempting to change the invoice number (this will raise an error)
try:
    invoice_metadata[0] = "INV-2026-9999B"
except TypeError as e:
    print("Error:", e)
```

### Dictionaries: Key-Value Pairs

A dictionary in Python is a mutable data structure that stores data in key-value pairs. Each key is unique and maps to a specific value, allowing for efficient data retrieval. Dictionaries are defined using curly braces `{}` containing key-value pairs separated by colons. A dictionary and can hold various data types as values, including other dictionaries, lists, or tuples.

Scenario: We can use a dictionary to represent the line items of the invoice, where each item is identified by its reference code. This allows us to easily access and manipulate the details of each product. In this example, we will create a dictionary where the keys are the reference codes of the line items, and the values are **nested dictionaries** containing the description, quantity, net weight, unit price, and total amount for each item.

### Sets: Unordered & Unique  

A set in Python is an unordered collection of **unique** elements. Sets are defined using curly braces `{}` or the `set()` constructor with elements separated by commas. They are useful for storing distinct items and performing operations like union, intersection, and difference. Sets do not allow duplicate values, making them ideal for scenarios where uniqueness is required. Order isn't really relevant to sets, but membership is. We use sets to check for the presence of specific member elements.  

Scenario: We can use a set to represent the certifications associated with the line items in the invoice. This allows us to easily check for the presence of specific certifications and ensure that all required certifications are included. Sets are excellent for compliance verification where you need to check for the presence of specific criteria without worrying about order or duplicate entries.

```python
certifications = {
    "Fairtrade",
    "RainForest Alliance",
    "Conventional"
}
# and a print statement to display the set
print("Certifications Set:", certifications)
# and check for the presence of a specific certification
print("Is 'Fairtrade' certification present?", "Fairtrade" in certifications)
# and check for the presence of a certification that is not in the set
print("Is 'Organic' certification present?", "Organic" in certifications)
```

Note that sets are unordered, so the order of elements may vary each time you print the set. Sets are also mutable, allowing you to add or remove elements as needed.

```python
# Adding a new certification
certifications.add("Organic")
# and a print statement to display the updated set
print("Updated Certifications Set:", certifications)
# Removing a certification
certifications.remove("Conventional")
# and a print statement to display the updated set
print("Updated Certifications Set after removal:", certifications)  
```
Also, because sets are unordered, you cannot access elements by index. If needed, you can iterate over the set to access its elements. 

```python
# Iterating over the set to print each certification
for certification in certifications:
    print(certification)
```

### Lists: Ordered & Mutable

A list in Python is an **ordered** collection of elements that can be of any data type. Lists are defined using square brackets `[]` and can hold multiple items separated by commas. Unlike tuples, lists are mutable, meaning that their contents can be changed after creation. Lists maintain the order of elements, allowing for indexing and slicing to access specific items.

Lists are usually taught as the first data collection structure in programming because they are easy to understand and use. They are versatile and can be used to store sequences of items, such as a list of products, prices, or any other related data. However, lists are so versatile that they can be misused and misunderstood, leading to confusion and errors in code. For example, using a list to store unique items or fixed identifiers can lead to unintended consequences, as lists allow duplicates and can be modified.

Scenario: We can use a list to represent the quantities of each line item in the invoice. This allows us to easily access and manipulate the quantities, as well as perform calculations such as total quantity or average quantity. Lists are ideal for scenarios where order matters and where we may need to modify the contents.

```python
quantities = [200, 300, 140]  # Quantities of line items
# and a print statement to display the list
print("Quantities List:", quantities)
# Accessing the quantity of the second line item (index 1)
print("Quantity of second line item:", quantities[1])
# Modifying the quantity of the first line item (index 0)
quantities[0] = 250
# and a print statement to display the updated list
print("Updated Quantities List:", quantities)
# Adding a new quantity for a new line item
quantities.append(100)
# and a print statement to display the updated list
print("Updated Quantities List after adding new item:", quantities)
# Removing a quantity (e.g., the last item)
quantities.pop()  # Removes the last item
# and a print statement to display the updated list
print("Updated Quantities List after removing last item:", quantities)
```
